```r


# sub() - replaces first occurrence
text <- "R Programming is easy. R Programming is useful."

print(sub("R Programming", "Python", text))


# gsub() - replaces all occurrences
print(gsub("R Programming", "Python", text))


# grep() - search matching strings
str <- c(
  "R Course",
  "Python Course",
  "R Programming",
  "Data Science"
)

print(grep("R", str, value = TRUE))


# grep() - return index numbers
print(grep("R", str, value = FALSE))


# grepl() - returns TRUE/FALSE
print(grepl("R", str))


# ---------------- DATA FRAME ----------------

students <- data.frame(
  Name = c("Amit", "Rahul", "Sneha", "Priya", "Karan"),
  Marks = c(85, 72, 91, 68, 77),
  Branch = c("CMPN", "IT", "CMPN", "IT", "CMPN")
)

# Display data frame
print(students)


# Display column names
print(colnames(students))


# Access columns
print(students$Name)
print(students$Marks)


# Check data types
print(is.numeric(students$Marks))
print(is.character(students$Name))


# Summary
print(summary(students))


# ---------------- SUBSET ----------------

# Select CMPN students
cmpn_students <- subset(
  students,
  Branch == "CMPN"
)

print(cmpn_students)


# Select students with marks >= 80
top_students <- subset(
  students,
  Marks >= 80
)

print(top_students)


# ---------------- SPLIT ----------------

# Split data according to Branch
split_data <- split(
  students,
  students$Branch
)

print(split_data)


# ---------------- CBIND ----------------

df1 <- data.frame(
  ID = c(1, 2, 3),
  Name = c("Amit", "Rahul", "Sneha")
)

df2 <- data.frame(
  Marks = c(85, 72, 91),
  Grade = c("A", "B", "A")
)

# Combine columns
combined_columns <- cbind(df1, df2)

print(combined_columns)


# ---------------- MERGE ----------------

df3 <- data.frame(
  ID = c(1, 2, 3),
  Branch = c("CMPN", "IT", "CMPN")
)

# Merge using common ID
merged_data <- merge(
  df1,
  df3,
  by = "ID"
)

print(merged_data)


# ---------------- RBIND ----------------

df4 <- data.frame(
  ID = c(4, 5),
  Name = c("Priya", "Karan")
)

# Combine rows
combined_rows <- rbind(
  df1,
  df4
)

print(combined_rows)


# ---------------- FINAL MESSAGE ----------------

print("NPTEL Week 10 R Programming completed successfully!")
```
