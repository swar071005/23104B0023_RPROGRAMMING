```r
# ============================================================
# FRS WEEK 05 - R PROGRAMMING
# Missing Data + Conditional Execution
# ============================================================


# ============================================================
# PART A: MISSING DATA HANDLING
# ============================================================

cat("\n========================================\n")
cat("1. MISSING DATA - NA\n")
cat("========================================\n")

x <- NA

print(x)
print(is.na(x))


# ------------------------------------------------------------
# 2. is.na() WITH VECTOR
# ------------------------------------------------------------

cat("\n========================================\n")
cat("2. is.na() WITH VECTOR\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

print(x)
print(is.na(x))


# ------------------------------------------------------------
# 3. MEAN WITH MISSING VALUES
# ------------------------------------------------------------

cat("\n========================================\n")
cat("3. MEAN WITH MISSING VALUES\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

cat("Mean without removing NA:\n")
print(mean(x))

cat("Mean after removing NA:\n")
print(mean(x, na.rm = TRUE))


# ------------------------------------------------------------
# 4. NA VERSUS NULL
# ------------------------------------------------------------

cat("\n========================================\n")
cat("4. NA AND NULL\n")
cat("========================================\n")

x <- NA
y <- NULL

cat("x = ")
print(x)

cat("y = ")
print(y)

cat("Is x NA?\n")
print(is.na(x))

cat("Is y NULL?\n")
print(is.null(y))


# ------------------------------------------------------------
# 5. LOCATION OF MISSING VALUES
# ------------------------------------------------------------

cat("\n========================================\n")
cat("5. LOCATION OF NA VALUES\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

print(x)

missing_positions <- which(is.na(x))

print(missing_positions)


# ------------------------------------------------------------
# 6. COUNT NUMBER OF NA VALUES
# ------------------------------------------------------------

cat("\n========================================\n")
cat("6. COUNT OF NA VALUES\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

number_of_missing <- sum(is.na(x))

print(number_of_missing)


# ------------------------------------------------------------
# 7. COMPLETE CASES
# ------------------------------------------------------------

cat("\n========================================\n")
cat("7. COMPLETE CASES\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

print(x)

complete <- complete.cases(x)

print(complete)


# ------------------------------------------------------------
# 8. REMOVE MISSING VALUES USING na.omit()
# ------------------------------------------------------------

cat("\n========================================\n")
cat("8. na.omit()\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

print(x)

y <- na.omit(x)

print(y)


# ------------------------------------------------------------
# 9. MEAN AFTER na.omit()
# ------------------------------------------------------------

cat("\n========================================\n")
cat("9. MEAN AFTER na.omit()\n")
cat("========================================\n")

x <- c(11, NA, 13, NA)

y <- na.omit(x)

cat("Original data:\n")
print(x)

cat("Data after removing NA:\n")
print(y)

cat("Mean:\n")
print(mean(y))


# ============================================================
# PART B: CONDITIONAL EXECUTION - IF
# ============================================================


# ------------------------------------------------------------
# 10. SIMPLE IF
# ------------------------------------------------------------

cat("\n========================================\n")
cat("10. SIMPLE IF\n")
cat("========================================\n")

x <- 5

if (x > 4) {
  print(x * 3)
}


# ------------------------------------------------------------
# 11. IF CONDITION FALSE
# ------------------------------------------------------------

cat("\n========================================\n")
cat("11. IF - FALSE CONDITION\n")
cat("========================================\n")

x <- 3

if (x > 4) {
  print(x * 3)
}


# ------------------------------------------------------------
# 12. IF WITH MESSAGE
# ------------------------------------------------------------

cat("\n========================================\n")
cat("12. IF WITH MESSAGE\n")
cat("========================================\n")

x <- 6

if (x > 3) {
  print("The value is more than 3")
}


# ------------------------------------------------------------
# 13. IF FALSE WITH MESSAGE
# ------------------------------------------------------------

cat("\n========================================\n")
cat("13. IF FALSE CONDITION\n")
cat("========================================\n")

x <- 2

if (x > 3) {
  print("The value is more than 3")
}


# ============================================================
# PART C: IF - ELSE
# ============================================================


# ------------------------------------------------------------
# 14. IF-ELSE: x = 5
# ------------------------------------------------------------

cat("\n========================================\n")
cat("14. IF-ELSE - x = 5\n")
cat("========================================\n")

x <- 5

if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}

print(x)


# ------------------------------------------------------------
# 15. IF-ELSE: x = 3
# ------------------------------------------------------------

cat("\n========================================\n")
cat("15. IF-ELSE - x = 3\n")
cat("========================================\n")

x <- 3

if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}

print(x)


# ------------------------------------------------------------
# 16. CHECK MORE OR LESS THAN 3
# ------------------------------------------------------------

cat("\n========================================\n")
cat("16. MORE OR LESS THAN 3\n")
cat("========================================\n")

x <- 6

if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}


# ------------------------------------------------------------
# 17. CHECK MORE OR LESS THAN 3 - x = 2
# ------------------------------------------------------------

cat("\n========================================\n")
cat("17. MORE OR LESS THAN 3 - x = 2\n")
cat("========================================\n")

x <- 2

if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}


# ============================================================
# PART D: NESTED IF - ELSE IF - ELSE
# ============================================================


# ------------------------------------------------------------
# 18. NESTED IF - x = 5
# ------------------------------------------------------------

cat("\n========================================\n")
cat("18. NESTED IF - x = 5\n")
cat("========================================\n")

x <- 5

if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}

print(x)


# ------------------------------------------------------------
# 19. NESTED IF - x = 2
# ------------------------------------------------------------

cat("\n========================================\n")
cat("19. NESTED IF - x = 2\n")
cat("========================================\n")

x <- 2

if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}

print(x)


# ------------------------------------------------------------
# 20. NESTED IF - x = 3
# ------------------------------------------------------------

cat("\n========================================\n")
cat("20. NESTED IF - x = 3\n")
cat("========================================\n")

x <- 3

if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}

print(x)


# ============================================================
# PART E: IFELSE()
# ============================================================


# ------------------------------------------------------------
# 21. ifelse() WITH VECTOR
# ------------------------------------------------------------

cat("\n========================================\n")
cat("21. ifelse() WITH VECTOR\n")
cat("========================================\n")

x <- 1:10

print(x)

result <- ifelse(x < 6, x^2, x + 1)

print(result)


# ------------------------------------------------------------
# 22. ifelse() FOR EVEN AND ODD
# ------------------------------------------------------------

cat("\n========================================\n")
cat("22. EVEN OR ODD USING ifelse()\n")
cat("========================================\n")

x <- c(7, 9, 8, 4)

print(x)

result <- ifelse(
  x %% 2 == 0,
  "even number",
  "odd number"
)

print(result)


# ============================================================
# PART F: SWITCH()
# ============================================================


# ------------------------------------------------------------
# 23. switch() WITH INTEGER
# ------------------------------------------------------------

cat("\n========================================\n")
cat("23. switch() - INTEGER\n")
cat("========================================\n")

print(switch(
  2,
  "apple",
  "banana",
  "orange"
))

print(switch(
  1,
  "apple",
  "banana",
  "orange"
))


# ------------------------------------------------------------
# 24. switch() WITH CHARACTER STRING
# ------------------------------------------------------------

cat("\n========================================\n")
cat("24. switch() - CHARACTER\n")
cat("========================================\n")

print(switch(
  "colour",
  colour = "blue",
  gender = "male",
  volume = 50
))

print(switch(
  "volume",
  colour = "blue",
  gender = "male",
  volume = 50
))


# ------------------------------------------------------------
# 25. switch() - NO MATCH
# ------------------------------------------------------------

cat("\n========================================\n")
cat("25. switch() - NO MATCH\n")
cat("========================================\n")

result1 <- switch(
  4,
  "apple",
  "banana",
  "orange"
)

print(result1)

result2 <- switch(
  "size",
  colour = "blue",
  gender = "male",
  volume = 50
)

print(result2)


# ============================================================
# PART G: which()
# ============================================================


# ------------------------------------------------------------
# 26. which() - EQUAL TO 14
# ------------------------------------------------------------

cat("\n========================================\n")
cat("26. which() - x == 14\n")
cat("========================================\n")

x <- c(10, 15, 8, 14, 6, 12)

print(x)

print(which(x == 14))


# ------------------------------------------------------------
# 27. which() - NOT EQUAL TO 12
# ------------------------------------------------------------

cat("\n========================================\n")
cat("27. which() - x != 12\n")
cat("========================================\n")

print(which(x != 12))


# ------------------------------------------------------------
# 28. which() - GREATER THAN 10
# ------------------------------------------------------------

cat("\n========================================\n")
cat("28. which() - x > 10\n")
cat("========================================\n")

print(which(x > 10))


# ============================================================
# PART H: which.min() AND which.max()
# ============================================================


# ------------------------------------------------------------
# 29. which.min() AND which.max()
# ------------------------------------------------------------

cat("\n========================================\n")
cat("29. which.min() AND which.max()\n")
cat("========================================\n")

x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

print(x)

cat("Position of minimum value:\n")
print(which.min(x))

cat("Position of maximum value:\n")
print(which.max(x))


# ============================================================
# PART I: which() WITH MATRIX
# ============================================================


# ------------------------------------------------------------
# 30. which() WITH ODD VALUES
# ------------------------------------------------------------

cat("\n========================================\n")
cat("30. which() - ODD VALUES\n")
cat("========================================\n")

x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

print(x)

print(which(x %% 2 == 1))


# ------------------------------------------------------------
# 31. which() WITH arr.ind = TRUE
# ------------------------------------------------------------

cat("\n========================================\n")
cat("31. which() WITH arr.ind = TRUE\n")
cat("========================================\n")

print(which(
  x %% 2 == 1,
  arr.ind = TRUE
))


# ============================================================
# END OF WEEK 05
# ============================================================

cat("\n")
cat("====================================================\n")
cat("       FRS WEEK 05 COMPLETED SUCCESSFULLY\n")
cat("====================================================\n")
```
