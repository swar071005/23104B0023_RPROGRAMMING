```r
# ============================================================
# R PROGRAMMING - NPTEL WEEK 11
# Complete Colab Code
# ============================================================

# ------------------------------------------------------------
# 1. DATA FRAME OPERATIONS
# ------------------------------------------------------------

painters <- data.frame(
  Composition = c(10, 15, 8, 12, 0),
  Drawing = c(8, 16, 13, 16, 15),
  Colour = c(16, 4, 16, 9, 8),
  Expression = c(3, 14, 7, 8, 0),
  School = c("A", "A", "A", "A", "A")
)

print(painters)

# Summary
summary(painters$School)
summary(painters$Composition)

# Attach
attach(painters)
summary(School)
summary(Composition)
detach(painters)

# Subset
print(subset(painters, School == "A"))
print(subset(painters, Composition <= 10))

# Select columns
print(subset(
  painters,
  School == "A",
  select = c(Composition, Drawing, Expression)
))

# Split data frame
splitted <- split(painters, painters$School)
print(splitted)

# Check data frame
print(is.data.frame(splitted$A))


# ------------------------------------------------------------
# 2. EXCEL FILE HANDLING
# ------------------------------------------------------------

# Install and load readxl
if (!require(readxl)) {
  install.packages("readxl")
  library(readxl)
}

# Create an example Excel-like dataset
# In Colab, upload your Excel file and use:
# data_excel <- read_excel("/content/datafile.xlsx")

# Example data frame
data_excel <- data.frame(
  Variable1 = c(1, 2, 3, 4, 5),
  Variable2 = c(10, 20, 30, 40, 50),
  Variable3 = c(100, 200, 300, 400, 500)
)

print(data_excel)

# Extract variables
print(data_excel$Variable1)
print(data_excel$Variable2)

# Mean
print(mean(data_excel$Variable1))


# ------------------------------------------------------------
# 3. CSV FILE HANDLING
# ------------------------------------------------------------

student <- data.frame(
  Name = c("Amit", "Rahul", "Swar", "Neha"),
  Marks = c(85, 78, 92, 88),
  Grade = c("A", "B", "A+", "A")
)

# Write CSV
write.csv(
  student,
  file = "/content/student.csv",
  row.names = FALSE
)

# Read CSV
student_read <- read.csv("/content/student.csv")

print(student_read)


# ------------------------------------------------------------
# 4. WRITE DATA TO FILE
# ------------------------------------------------------------

x <- 1:100

write(
  x,
  file = "/content/shalabh.txt"
)

print("Data successfully written to file.")


# ------------------------------------------------------------
# 5. ABSOLUTE AND RELATIVE FREQUENCY
# ------------------------------------------------------------

gender <- c(
  1, 2, 1, 2, 1,
  1, 1, 2, 1, 1
)

# Absolute frequency
absolute_frequency <- table(gender)

print("Absolute Frequency:")
print(absolute_frequency)

# Relative frequency
relative_frequency <- table(gender) / length(gender)

print("Relative Frequency:")
print(relative_frequency)


# ------------------------------------------------------------
# 6. PIZZA DELIVERY FREQUENCY
# ------------------------------------------------------------

direction <- c(
  1,1,2,1,2,3,2,2,3,3,3,1,2,3,2,2,3,1,1,3,
  3,1,2,1,3,3,3,2,2,2,2,1,2,2,1,1,1,3,2,2,1,
  2,3,2,2,1,2,3,3,2,1,2,2,3,1,1,2,1,2,3,2,3,
  2,2,3,1,2,3,3,3,2,1,1,1,2,1,1,2,1,2,3,3,1,
  2,3,3,2,1,2,3,2,1,3,2,2,2,2,3,2,2
)

# Absolute frequency
print("Direction Absolute Frequency:")
print(table(direction))

# Relative frequency
print("Direction Relative Frequency:")
print(table(direction) / length(direction))


# ------------------------------------------------------------
# 7. QUANTILES
# ------------------------------------------------------------

marks <- c(
  68, 82, 63, 86, 34,
  96, 41, 89, 29, 51,
  75, 77, 56, 59, 42
)

# Default quantiles
print("Default Quantiles:")
print(quantile(marks))

# 0%, 25%, 50%, 75%, 100%
print("Standard Quantiles:")
print(
  quantile(
    marks,
    probs = c(0, 0.25, 0.5, 0.75, 1)
  )
)

# Custom probabilities
print("Custom Quantiles:")
print(
  quantile(
    marks,
    probs = c(0, 0.20, 0.40, 0.60, 0.80, 1)
  )
)


# ------------------------------------------------------------
# 8. SCATTER PLOT
# ------------------------------------------------------------

height <- c(
  166,125,130,142,147,159,159,147,165,156,
  149,164,137,166,135,142,133,136,127,143,
  165,121,142,148,158,146,154,157,124,125,
  158,159,164,143,154,152,141,164,131,152,
  152,161,143,143,139,131,125,145,140,163
)

plot(
  height,
  main = "Height of 50 Persons",
  xlab = "Person Number",
  ylab = "Height (cm)",
  pch = 19
)


# ------------------------------------------------------------
# 9. SCATTER PLOT BETWEEN TWO VARIABLES
# ------------------------------------------------------------

x <- 1:10
y <- c(12, 15, 14, 18, 20, 19, 23, 25, 24, 28)

plot(
  x,
  y,
  main = "Scatter Plot",
  xlab = "X",
  ylab = "Y",
  pch = 19
)


# ------------------------------------------------------------
# 10. BAR PLOT - ABSOLUTE FREQUENCY
# ------------------------------------------------------------

barplot(
  table(gender),
  main = "Gender Frequency",
  xlab = "Gender",
  ylab = "Frequency"
)


# ------------------------------------------------------------
# 11. BAR PLOT - RELATIVE FREQUENCY
# ------------------------------------------------------------

barplot(
  table(gender) / length(gender),
  main = "Relative Frequency of Gender",
  xlab = "Gender",
  ylab = "Relative Frequency"
)


# ------------------------------------------------------------
# 12. BAR PLOT - PIZZA DELIVERY
# ------------------------------------------------------------

barplot(
  table(direction),
  main = "Directions of Food Delivery",
  xlab = "Food Delivery Directions",
  ylab = "Number of Deliveries"
)


# ------------------------------------------------------------
# 13. BAR PLOT WITH TITLE, LEGEND AND SUBTITLE
# ------------------------------------------------------------

barplot(
  table(direction),
  main = "Directions of Food Delivery",
  sub = "Three Directions",
  xlab = "Food Delivery Directions",
  ylab = "Number of Deliveries",
  legend.text = c("East", "West", "Central")
)

```
