```r


# 1. Basic calculations
print(10 + 5)
print(10 - 5)
print(10 * 5)
print(10 / 5)

# 2. Assignment
x <- 20
y <- 10
print(x + y)

# 3. Vectors
x <- c(2, 3, 5, 7)
y <- c(-2, -3, -5, 8)

print(x + y)
print(x - y)
print(x * y)
print(x / y)

# 4. Different length vectors
x <- c(2, 3, 5, 7)
y <- c(8, 9)

print(x + y)
print(x - y)
print(x * y)
print(x / y)

# 5. Numeric and character
x <- 20
y <- "apple"

print(is.numeric(x))
print(is.character(y))

# 6. Type conversion
x <- 20
y <- as.character(x)
print(y)

z <- as.numeric(y)
print(z)

# 7. Mode
print(mode(x))
print(mode(y))

# 8. Storage mode
print(storage.mode(x))
print(storage.mode(y))

# 9. Infinity
print(3 / 0)
print(is.infinite(3 / 0))
print(is.finite(10))

# 10. Scalar operations
x <- c(2, 3, 5, 7)

print(x + 10)
print(x - 10)
print(x * 10)
print(x / 10)

# 11. BODMAS
print((2 + 3) * 5 + 5 - 10)

# 12. Comments
# This is a comment

#
```
