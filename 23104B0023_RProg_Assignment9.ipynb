```r
# ==========================================
# R PROGRAMMING - WEEK 9
# STRING OPERATIONS
# ==========================================

# ==========================================
# 1. paste()
# ==========================================

paste(
  "Everybody",
  "loves",
  "R Programming."
)


# ==========================================
# 2. paste() with separator
# ==========================================

paste(
  "Everybody",
  "loves",
  "R Programming.",
  sep = "*"
)


# ==========================================
# 3. paste() with custom separator
# ==========================================

paste(
  "Everybody",
  "loves",
  "R Programming.",
  sep = "==="
)


# ==========================================
# 4. paste() with vectors
# ==========================================

names <- c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)

paste(
  names,
  "is",
  "a good",
  "person."
)


# ==========================================
# 5. paste0()
# ==========================================

paste0(
  "R",
  "Programming"
)


# ==========================================
# 6. paste0() with vector
# ==========================================

x <- c(
  "A",
  "B",
  "C"
)

paste0(
  x,
  "123"
)


# ==========================================
# 7. collapse
# ==========================================

paste(
  c(
    "R",
    "Python",
    "Java"
  ),
  collapse = ", "
)


# ==========================================
# 8. paste with sep and collapse
# ==========================================

paste(
  c("A", "B", "C"),
  c("1", "2", "3"),
  sep = "-",
  collapse = ", "
)


# ==========================================
# 9. strsplit()
# ==========================================

x <- "The&!syntax&!of&!paste"

strsplit(
  x,
  split = "!"
)


# ==========================================
# 10. Split using multiple characters
# ==========================================

x <- "The&!syntax&!of&!paste&!is"

strsplit(
  x,
  split = "&!"
)


# ==========================================
# 11. Access components of split result
# ==========================================

y <- strsplit(
  x,
  split = "&!"
)

y[[1]][1]

y[[1]][2]

y[[1]][3]


# ==========================================
# 12. Split dates
# ==========================================

dates <- c(
  "2020-07-24",
  "2021-08-25",
  "2022-09-26",
  "2023-10-27"
)

dates_split <- strsplit(
  dates,
  split = "-"
)

dates_split


# ==========================================
# 13. Extract year
# ==========================================

dates_split[[1]][1]

dates_split[[2]][1]

dates_split[[3]][1]

dates_split[[4]][1]


# ==========================================
# 14. Extract month
# ==========================================

dates_split[[1]][2]

dates_split[[2]][2]

dates_split[[3]][2]

dates_split[[4]][2]


# ==========================================
# 15. Extract day
# ==========================================

dates_split[[1]][3]

dates_split[[2]][3]

dates_split[[3]][3]

dates_split[[4]][3]


# ==========================================
# 16. Convert split dates to matrix
# ==========================================

date_matrix <- do.call(
  rbind,
  dates_split
)

date_matrix


# ==========================================
# 17. String concatenation example
# ==========================================

first_name <- c(
  "Swar",
  "Rahul",
  "Amit"
)

last_name <- c(
  "Sawalkar",
  "Patil",
  "Sharma"
)

paste(
  first_name,
  last_name
)


# ==========================================
# 18. Student ID creation
# ==========================================

roll <- c(
  "23104",
  "23105",
  "23106"
)

paste0(
  "CMPN-",
  roll
)
```
